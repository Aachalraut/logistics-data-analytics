# Week 3 — Advanced Data Analysis and Visualization
This notebook performs EDA on a simulated logistics dataset and produces publication-ready PNG visualizations.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/raw/week3_week4_logistics_simulated.csv')
df.head()
```

```python
df.describe().round(2)
```

```python
df.select_dtypes('number').corr().round(2)
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['delivery_time_hours'], bins=25)
plt.title('Distribution of Delivery Time')
plt.xlabel('Delivery Time (hours)')
plt.ylabel('Shipments')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['distance_km'], df['delivery_time_hours'], alpha=.55)
plt.title('Distance vs Delivery Time')
plt.xlabel('Distance (km)')
plt.ylabel('Delivery Time (hours)')
plt.show()
```
